### Growth rates retransformation

1. $ growth\_rate = (present / past) ** (1 / n) - 1 $
2. $ present = past * (1 + growth\_rate) ** n $

```python
step = 3
n = step / 12
```

```python
# Actual growth rates calculation & retransformation
idx = pd.concat([y_train, y_test]).resample("QS").last().index
past = Z_df[target_col_name].loc[idx].shift(1)

present = Z_df[target_col_name].loc[idx]
actual_growth_rates = (present / past).dropna() ** (1 / n) - 1

print(past * (1 + actual_growth_rates) ** n)  # present values – estimated based on raw growth rated
```

```python
# Smooth growth rates retransformation
smooth_growth_rates = pd.concat([y_train, y_test]).resample("QS").mean()/100
idx = smooth_growth_rates.index

past = Z_df[[target_col_name]].loc[idx].shift(1)

# smooth_growth_rates = y_train.resample("QS").last().shift(1)/100
past * (1 + smooth_growth_rates) ** n  #  present values – estimated based on smoothed growth rated
```
